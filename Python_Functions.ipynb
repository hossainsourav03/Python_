#### Nested Function

```python
def f():
    print("Inside f")
    def g():
        print("Inside g")
    g()

f()
```

```python
g()
```

```python
def g(x):
    def h():
        x = "abc"
    x = x+1
    print("in g(x): x = ", x)
    h()
    return x

x = 3
z = g(x)
```

```python
def g(x):
    def h(x):
        x = x + 1
        print("in h(x): x = ", x)
    x = x+1
    print("in g(x): x = ", x)
    h(x)
    return x

x = 3
z = g(x)
print("in main code: x = ", x)
print("in main code: z = ", z)
```
