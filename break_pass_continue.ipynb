#### Break
for breaking the loop means full stop the loop where we call the break there will be not a single iteration 

```python
for i in range(1, 11):
    if i == 5:
        break
    print(i)
```

#### Continue
will skip the point where we call the continue

```python
for i in range(1, 11):
    if i == 5:
        continue
    print(i)
```

#### Pass
for use to jump into another part and come back again to that specific segment where we use the pass statement.

```python
for i in range(1, 11):
    pass
```
