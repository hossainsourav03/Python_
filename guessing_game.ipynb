```python
import random
```

```python
jackpot = random.randint(1, 100)
```

```python
number = int(input("Enter a number: "))
i = 1

while number != jackpot:
    if number < jackpot:
        print("Guess Higher")
    else:
        print("Guess Lower")
    number = int(input("Enter a number again: "))
    i += 1
print("Congratulation!!!!You guess right number.")
print("Total guessing ", i)
```
