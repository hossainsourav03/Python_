#### Mult Function

```python
def mul(a, b):
    if b == 1:
        return a
    else:
        result = a + mul(a, b-1)
    return result

mul(3, 4)
```

#### Factorial

```python
def fact(n):
    if n == 1:
        return n
    else:
        return n * fact(n-1)
print(fact(5))
```

#### Palindrome

```python
def pal(s):
    if len(s) <= 1:
        return "palindrome"
    else:
        if s[0] == s[-1]:
            return pal(s[1:-1])
        else:
            return "Not a palindrome"
print(pal("amma"))    
```

#### Fibonacci

```python
def fibo(n):
    if n == 0 or n == 1:
        return 1
    else:
        return fibo(n-1) + fibo(n-2)

print(fibo(12))
```

```python
import time
def fibo(n):
    if n == 0 or n == 1:
        return 1
    else:
        return fibo(n-1) + fibo(n-2)

start = time.time()
print(fibo(41))
print(time.time() - start)
```

```python
import time
def fibo(n, d):
    if n in d:
        return d[n]
    else:
        d[n] = fibo(n-1, d) + fibo(n-2, d)
        return d[n]

d = {0: 1, 1: 1}
start = time.time()
print(fibo(41, d))
print(time.time() - start)
```

#### Superset

```python
def super_set(l):
    if len(l) == 0:
        return [[]]
    first = l[0]
    rest = super_set(l[1:])
    result = []

    for sublist in rest:
        result.append(sublist)
        result.append([first] + sublist)
    return result

print(super_set([1, 2, 3]))
```
