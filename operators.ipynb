#### Identical Operator

Identical operator means is that two different variable share same memory location or not

```python
a = 3
b = 3
print(a is b)
```

```python
a = "Hello"
b = "Hello"
print(a is b)
```

```python
a = [1, 2, 3]
b = [1, 2, 3]
print(a is b)
```

```python
a = [1, 2, 3]
b = [1, 2, 3]
print(a is b)
```

```python
a = "Hello - world"
b = "Hello - world"
print(a is b)
```

```python
a = "Hello - world"
b = "Hello - world"
print(a is not b)
```
