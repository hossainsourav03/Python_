### Literals
there are 4 types of literals in python
##### 1 is numerical
##### 2 is string
##### 3 is boolean
##### 4 is special

#### Numericals

```python
#Numericals
a = 100 # decimal
b = 0b1010 #binary
c = 0o451 #octal
d = 0x5c0 #hexadecimal

print(a, b, c, d)

#float
a1 = 1.45
b1 = 1.5e2
c1 = 1.5e-2
print(a1, b1, c1)

#complex number
x = 0 + 1.7j
print(x, x.imag, x.real)
```

```python
#String Literals

string1 = 'My name is sourav.'
string2 = "My name is Sourav"
string3 = "C"
String4 = """My name is \nSyed Towhid Hossain Sourav.
And I am 23 years old."""
print(String4)

string5 = r"Computer Science \n and Engineering"
print(string5)
```

```python
#Boolean literals
a = True + 5
b = False + 10
print("a: ", a)
print("b: ", b)
```

### Special Literals

```python
z = None
print(z)
```
